```python
from pathlib import Path 
import json
import os

def _read_json(path: str) -> dict:
    """Read a JSON file and return its contents as a dict."""
    try:
        with open(path) as f:
            return json.load(f)
    except FileNotFoundError as e:
        raise FileNotFoundError(f"Required file not found: {path}") from e
    except json.JSONDecodeError as e:
        raise ValueError(f"Invalid JSON format in file: {path}") from e
    
main_path = Path(r"C:/Data/tests")

rewrite = False # Set to False to skip analysis if summary.json already exists in session folder

dirs = [os.path.join(main_path, name) for name in os.listdir(main_path) if os.path.isdir(os.path.join(main_path, name))]
print(dirs)

results = {}
for path in dirs:
    rig_data = _read_json(Path(os.path.join(path, "behavior" ,"Logs", "rig_input.json")))

    rig_info = {}  # initialize as dict
    rig_info["computer_name"] = rig_data["computer_name"]
    rig_info["rig_name"] = rig_data["rig_name"]
    # calibration': {'water_valve': {'device_name': 'WaterValve', 'input': 
    rig_info["water-cal"] = rig_data["calibration"]["water_valve"]["output"]
#    print(rig_data)
    print(rig_info["rig_name"], "/", rig_info["water-cal"]["slope"], "/", rig_info["water-cal"]["offset"], "/", rig_info["water-cal"]["r2"], "/", rig_info["water-cal"])

```
